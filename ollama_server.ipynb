This notebook runs Ollama on a free Google Colab GPU (T4/V100) and exposes it over an **ngrok** tunnel.
This allows your Job Hunting Agent on your Windows laptop to use Colab's cloud GPU for ultra-fast completions without consuming local battery or RAM.

### Step 2: Start the Ollama Service in Background

```python
import subprocess, time
process = subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(5)
print("✅ Ollama server is running on localhost:11434!")
```

### Step 4: Expose Port 11434 via Ngrok Tunnel
1. Create a free account at [https://ngrok.com](https://ngrok.com)
2. Copy your Auth Token from [https://dashboard.ngrok.com/get-started/your-authtoken](https://dashboard.ngrok.com/get-started/your-authtoken)
3. Paste it below and run the cell.

```python
!pip install pyngrok -q
from pyngrok import ngrok

# ⚠️ PASTE YOUR NGROK AUTH TOKEN HERE
NGROK_AUTH_TOKEN = "YOUR_NGROK_AUTH_TOKEN_HERE"

ngrok.set_auth_token(NGROK_AUTH_TOKEN)
tunnel = ngrok.connect(11434)
print("=" * 60)
print(f"🌐 Your Public Ollama URL: {tunnel.public_url}")
print("⚠️ Copy this URL and set OLLAMA_BASE_URL in your Windows .env file!")
print("=" * 60)
```

### Step 5: Keep-Alive Loop (Prevents Colab Idle Disconnect)

```python
import time, datetime
print("🔄 Keep-alive heartbeat started. Leave this tab open...")
while True:
    now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{now}] Heartbeat: Ollama server active on Colab")
    # Small command to keep CPU/GPU active
    !ollama list > /dev/null
    time.sleep(180)  # Every 3 minutes
```
